```python
import polars as pl
print(pl.__version__)
```

```python
from huggingface_hub import hf_hub_download

dataset_path = hf_hub_download(
    repo_id="NextGig-Rocks/global-job-postings-multi-ats",
    filename="nextgig_jobs_2026-06.parquet",
    repo_type="dataset",
)

print(dataset_path)
```

```python
df = pl.read_parquet(dataset_path)
print(df.shape)
print(df.columns)
print(df.schema)
```

## Occupational Categories and their amount

```python
job_offers = df['occupational_category'].value_counts()
print(job_offers)
```

## Number of columns

```python
count_columns = len(df.columns)
print(count_columns)
```

## Data Types

```python
types = df.dtypes
print(types)

df_tipos = pl.DataFrame({"tipos": [str(t) for t in df.dtypes]})
print(df_tipos["tipos"].value_counts())
```

```python
count_nulls = df.null_count()
print(count_nulls)
```

```python
countries = df['country'].value_counts()
print(countries)
```

```python
cities = df['city'].value_counts()
print(cities)
```

```python
experience = df['experience_level'].value_counts()
print(experience)
```

```python
work_model = df['work_model'].value_counts()
print(work_model)
```

```python
nulls_percent = df.select(
    pl.col('work_model').value_counts(normalize=True).struct.field("*")
).with_columns(
    (pl.col("proportion") * 100).alias("percent")
)

print(nulls_percent)
```
